```python
import duckdb
import pandas as pd

EMBEDDINGS_FILE = "app_duckdb\data\embeddings.duckdb"

search_phrase = "Fake sincerity" # Substitua pela palavra que você quer pesquisar

with duckdb.connect(EMBEDDINGS_FILE) as conn:
    result = conn.execute(f"""
        SELECT id, text
        FROM embeddings
        WHERE text LIKE '%{search_phrase}%'
        LIMIT 5
    """).fetchdf()

print(f"\nComentários contendo '{search_phrase}':")
texts_series = result['text']
for text in texts_series:
    print('\n')
    print(text)
```
